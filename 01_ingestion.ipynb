```python
%sql
-- table creation code
CREATE TABLE IF NOT EXISTS fk_1374.bronze.stores;
CREATE TABLE IF NOT EXISTS fk_1374.bronze.customers;
CREATE TABLE IF NOT EXISTS fk_1374.bronze.products;
```

```python
%sql
-- load all stores into bronze with audit columns
COPY INTO fk_1374.bronze.stores
FROM (
  SELECT *, _metadata.file_path AS _source_file, current_timestamp() AS _ingestion_at
  FROM '/Volumes/fk_1374/landing/raw/stores/'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');
```

```python
%sql
-- load all customers into bronze with audit columns
COPY INTO fk_1374.bronze.customers
FROM (
  SELECT *, _metadata.file_path AS _source_file, current_timestamp() AS _ingestion_at
  FROM '/Volumes/fk_1374/landing/raw/customers/'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');
```

```python
%sql
-- load all products into bronze with audit columns (nested JSON / array of JSON)
COPY INTO fk_1374.bronze.products
FROM (
  SELECT *, _metadata.file_path AS _source_file, current_timestamp() AS _ingestion_at
  FROM '/Volumes/fk_1374/landing/raw/products/'
)
FILEFORMAT = JSON
FORMAT_OPTIONS ('multiLine' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');
```
